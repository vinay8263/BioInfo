####**TASK P13.2** Explore file object functions to read and print practice.txt

````python
from google.colab import files


uploaded = files.upload()
filename = next(iter(uploaded))

print("\n--- Reading first line with readline() ---")
with open(filename, "r") as file:
    first_line = file.readline()
    print(first_line)

print("\n--- Reading all lines into a list with readlines() ---")
with open(filename, "r") as file:
    all_lines = file.readlines()
    # Print the first few lines to avoid flooding the output if the file is very large
    for i, line in enumerate(all_lines):
        if i < 5:
            print(line, end='') # end='' to prevent double newlines as lines already contain them
        else:
            break
    if len(all_lines) > 5:
        print("... (truncated, total lines: {}) ".format(len(all_lines)))
````

####**TASK P18.2** <font color= orange>**Complete Rosalind Problem #5  Dictionaries**</font>

````python
from google.colab import files

# Upload the INI6 dataset and build a list of case-sensitive words.
dataset_files = files.upload()
dataset_name = next(iter(dataset_files))

with open(dataset_name, "r") as input_file:
    text_words = input_file.read().split()

# Create each dictionary entry once, then increase it for every occurrence.
word_frequencies = {}
for current_word in text_words:
    if current_word not in word_frequencies:
        word_frequencies[current_word] = 0
    word_frequencies[current_word] += 1

# Each key and its final frequency are printed on their own line.
for current_word in word_frequencies:
    print(current_word, word_frequencies[current_word])
````

####**TASK P19.2** Modify your code for the Rosalind #5 Dictionary problem so that it uses count( ) to count the words.

````python
from google.colab import files

# Upload the dictionary dataset and read its words.
dataset_files = files.upload()
word_list = next(iter(dataset_files.values())).decode("utf-8").split()

# Preserve the first appearance of each word before using count().
unique_words = list(dict.fromkeys(word_list))
count_table = {
    current_word: word_list.count(current_word)
    for current_word in unique_words
}

for current_word, occurrence_total in count_table.items():
    print(current_word, occurrence_total)
````

#<font color=blue>**P23.**</font> SeqUtils.gc_fraction( ) – a function to compute GC content

####**TASK P23.1** Modify the script from TASK P22.1 so that it prints the GC content for each sequence in the file

````python
from google.colab import files
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction

# Upload the FASTA dataset used for the GC calculation.
fasta_files = files.upload()
fasta_name = next(iter(fasta_files))

# Convert each GC fraction to a percentage before displaying it.
with open(fasta_name, "r") as fasta_file:
    for fasta_record in SeqIO.parse(fasta_file, "fasta"):
        gc_percentage = gc_fraction(fasta_record.seq) * 100
        print(fasta_record.id, f"{gc_percentage:.6f}")
````

####**TASK P23.2** <font color=orange> **Complete Rosalind #9  Computing GC Content**</font>

````python
from google.colab import files
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction

# Upload the FASTA dataset containing the candidate DNA strings.
fasta_files = files.upload()
fasta_name = next(iter(fasta_files))

with open(fasta_name, "r") as fasta_file:
    fasta_records = list(SeqIO.parse(fasta_file, "fasta"))

# Select the record with the largest GC fraction.
top_record = max(fasta_records, key=lambda record: gc_fraction(record.seq))
top_gc_percentage = gc_fraction(top_record.seq) * 100

print(top_record.id)
print(f"{top_gc_percentage:.6f}")
````
